```python
!pip install shap -q

import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import StratifiedKFold, cross_val_score, train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier, export_text, plot_tree
from sklearn.metrics import (
    classification_report, confusion_matrix, f1_score,
    accuracy_score, precision_score, recall_score,
    ConfusionMatrixDisplay, make_scorer
)
import shap
import os
import glob

# Create output directory
os.makedirs('/kaggle/working/outputs', exist_ok=True)
OUT = '/kaggle/working/outputs'

# ============================================================
# 1. DATA LOADING
# ============================================================
print("=" * 70)
print("SECTION 1: DATA LOADING")
print("=" * 70)

df_mat = pd.read_csv('/kaggle/input/datasets/coolboyw3r21/student-dataset/student-mat.csv', sep=';')
df_por = pd.read_csv('/kaggle/input/datasets/coolboyw3r21/student-dataset/student-por.csv', sep=';')
print(f"Mathematics dataset loaded: {df_mat.shape}")
print(f"Portuguese dataset loaded:  {df_por.shape}")

df_mat['subject'] = 'Mathematics'
df_por['subject'] = 'Portuguese'
df = pd.concat([df_mat, df_por], ignore_index=True)
print(f"\nCombined dataset shape: {df.shape}")
print(f"Columns: {list(df.columns)}")

# ============================================================
# 2. EXPLORATORY DATA ANALYSIS
# ============================================================
print("\n" + "=" * 70)
print("SECTION 2: EXPLORATORY DATA ANALYSIS")
print("=" * 70)

print("\nFirst 5 rows:")
print(df.head())

print("\nDataset Info:")
print(f"Total samples: {len(df)}")
print(f"Total features: {df.shape[1]}")
print(f"Missing values per column:")
print(df.isnull().sum()[df.isnull().sum() > 0] if df.isnull().sum().sum() > 0 else "None")

print("\nDescriptive Statistics (numeric columns):")
print(df.describe().round(2))

df['outcome'] = (df['G3'] >= 10).astype(int)
print(f"\nTarget variable distribution:")
print(df['outcome'].value_counts())
print(f"Pass rate: {df['outcome'].mean():.3f}")

# ---- Figure 1: Target Distribution ----
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
df['outcome'].value_counts().plot(kind='bar', ax=axes[0], color=['#e74c3c', '#2ecc71'])
axes[0].set_title('Target Distribution (Pass/Fail)', fontsize=14, fontweight='bold')
axes[0].set_xticklabels(['Pass (1)', 'Fail (0)'], rotation=0)
axes[0].set_ylabel('Count')
for i, v in enumerate(df['outcome'].value_counts().values):
    axes[0].text(i, v + 5, str(v), ha='center', fontweight='bold')

df.groupby('subject')['outcome'].value_counts(normalize=True).unstack().plot(
    kind='bar', ax=axes[1], color=['#e74c3c', '#2ecc71']
)
axes[1].set_title('Pass Rate by Subject', fontsize=14, fontweight='bold')
axes[1].set_ylabel('Proportion')
axes[1].set_xticklabels(['Mathematics', 'Portuguese'], rotation=0)
axes[1].legend(['Fail', 'Pass'])
plt.tight_layout()
plt.savefig(f'{OUT}/fig1_target_distribution.png', dpi=150, bbox_inches='tight')
plt.show()

# ---- Figure 2: Grade Distribution ----
fig, ax = plt.subplots(figsize=(10, 5))
for subj in ['Mathematics', 'Portuguese']:
    subset = df[df['subject'] == subj]
    ax.hist(subset['G3'], bins=21, alpha=0.6, label=subj, edgecolor='black')
ax.axvline(x=10, color='red', linestyle='--', linewidth=2, label='Pass Threshold (10)')
ax.set_title('Distribution of Final Grades (G3) by Subject', fontsize=14, fontweight='bold')
ax.set_xlabel('Final Grade (G3)')
ax.set_ylabel('Frequency')
ax.legend()
plt.tight_layout()
plt.savefig(f'{OUT}/fig2_grade_distribution.png', dpi=150, bbox_inches='tight')
plt.show()

# ============================================================
# 3. FEATURE ENGINEERING & PREPROCESSING
# ============================================================
print("\n" + "=" * 70)
print("SECTION 3: FEATURE ENGINEERING & PREPROCESSING")
print("=" * 70)

features_to_drop = ['G1', 'G2', 'G3', 'outcome', 'subject']
X = df.drop(columns=features_to_drop)
y = df['outcome']

print(f"\nFeatures used (pre-outcome only): {list(X.columns)}")
print(f"Number of features: {X.shape[1]}")

cat_cols = X.select_dtypes(include=['object']).columns.tolist()
num_cols = X.select_dtypes(include=['int64', 'float64']).columns.tolist()
print(f"Categorical features ({len(cat_cols)}): {cat_cols}")
print(f"Numerical features ({len(num_cols)}): {num_cols}")

le_dict = {}
X_encoded = X.copy()
for col in cat_cols:
    le = LabelEncoder()
    X_encoded[col] = le.fit_transform(X_encoded[col])
    le_dict[col] = le
    print(f"  Encoded '{col}': {dict(zip(le.classes_, le.transform(le.classes_)))}")

print(f"\nEncoded feature matrix shape: {X_encoded.shape}")
print(f"\nClass balance:\n  Pass: {y.sum()} ({y.mean()*100:.1f}%)\n  Fail: {(1-y).sum()} ({(1-y.mean())*100:.1f}%)")

# ---- Figure 3: Correlation Heatmap ----
fig, ax = plt.subplots(figsize=(16, 12))
corr = X_encoded.copy()
corr['outcome'] = y.values
corr_matrix = corr.corr()
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
sns.heatmap(corr_matrix, mask=mask, annot=False, cmap='RdBu_r', center=0,
            ax=ax, square=True, linewidths=0.5)
ax.set_title('Feature Correlation Heatmap', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig(f'{OUT}/fig3_correlation_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()

outcome_corr = corr_matrix['outcome'].drop('outcome').abs().sort_values(ascending=False)
print("\nTop 10 features correlated with outcome:")
for feat, val in outcome_corr.head(10).items():
    print(f"  {feat}: {val:.3f}")

# ============================================================
# 4. TRAIN-TEST SPLIT
# ============================================================
print("\n" + "=" * 70)
print("SECTION 4: TRAIN-TEST SPLIT")
print("=" * 70)

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42, stratify=y
)
print(f"Training set: {X_train.shape[0]} samples")
print(f"Test set:     {X_test.shape[0]} samples")
print(f"Train class balance: Pass={y_train.sum()}, Fail={(1-y_train).sum()}")
print(f"Test class balance:  Pass={y_test.sum()}, Fail={(1-y_test).sum()}")

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# ============================================================
# 5. MODEL TRAINING AND EVALUATION
# ============================================================
print("\n" + "=" * 70)
print("SECTION 5: MODEL TRAINING AND EVALUATION")
print("=" * 70)

print("\n--- 5a. Dummy Classifier (Most Frequent Strategy) ---")
dummy = DummyClassifier(strategy='most_frequent', random_state=42)
dummy.fit(X_train, y_train)
y_pred_dummy = dummy.predict(X_test)
print(f"Dummy always predicts: {dummy.class_prior_}")
print(classification_report(y_test, y_pred_dummy, target_names=['Fail', 'Pass']))
macro_f1_dummy = f1_score(y_test, y_pred_dummy, average='macro')
print(f"Macro F1 (Dummy): {macro_f1_dummy:.4f}")

print("\n--- 5b. Logistic Regression ---")
lr = LogisticRegression(max_iter=1000, random_state=42, solver='lbfgs')
lr.fit(X_train_scaled, y_train)
y_pred_lr = lr.predict(X_test_scaled)
print(classification_report(y_test, y_pred_lr, target_names=['Fail', 'Pass']))
macro_f1_lr = f1_score(y_test, y_pred_lr, average='macro')
print(f"Macro F1 (Logistic Regression): {macro_f1_lr:.4f}")

print("\n--- 5c. Decision Tree Classifier ---")
dt = DecisionTreeClassifier(
    max_depth=5, min_samples_split=10, min_samples_leaf=5,
    random_state=42, class_weight='balanced'
)
dt.fit(X_train, y_train)
y_pred_dt = dt.predict(X_test)
print(classification_report(y_test, y_pred_dt, target_names=['Fail', 'Pass']))
macro_f1_dt = f1_score(y_test, y_pred_dt, average='macro')
print(f"Macro F1 (Decision Tree): {macro_f1_dt:.4f}")

# ============================================================
# 6. CROSS-VALIDATION
# ============================================================
print("\n" + "=" * 70)
print("SECTION 6: STRATIFIED 5-FOLD CROSS-VALIDATION")
print("=" * 70)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
macro_f1_scorer = make_scorer(f1_score, average='macro')

models = {
    'Dummy (Most Frequent)': DummyClassifier(strategy='most_frequent', random_state=42),
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42, solver='lbfgs'),
    'Decision Tree': DecisionTreeClassifier(max_depth=5, min_samples_split=10,
                                             min_samples_leaf=5, random_state=42,
                                             class_weight='balanced')
}

cv_results = {}
for name, model in models.items():
    if name == 'Logistic Regression':
        pipe_X = pd.DataFrame(scaler.fit_transform(X_encoded), columns=X_encoded.columns)
        scores = cross_val_score(model, pipe_X, y, cv=cv, scoring=macro_f1_scorer)
    else:
        scores = cross_val_score(model, X_encoded, y, cv=cv, scoring=macro_f1_scorer)
    cv_results[name] = scores
    print(f"{name}:")
    print(f"  Fold scores: {scores.round(4)}")
    print(f"  Mean Macro F1: {scores.mean():.4f} (+/- {scores.std():.4f})")

# ---- Figure 4: Model Comparison Bar Chart ----
fig, ax = plt.subplots(figsize=(10, 6))
model_names = list(cv_results.keys())
means = [cv_results[m].mean() for m in model_names]
stds = [cv_results[m].std() for m in model_names]
colors = ['#95a5a6', '#3498db', '#2ecc71']
bars = ax.bar(model_names, means, yerr=stds, capsize=5, color=colors, edgecolor='black')
ax.set_ylabel('Macro F1 Score', fontsize=12)
ax.set_title('Model Comparison: Macro F1 (5-Fold CV)', fontsize=14, fontweight='bold')
ax.set_ylim(0, 1)
for bar, mean, std in zip(bars, means, stds):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + std + 0.02,
            f'{mean:.3f}', ha='center', fontweight='bold', fontsize=12)
plt.tight_layout()
plt.savefig(f'{OUT}/fig4_model_comparison.png', dpi=150, bbox_inches='tight')
plt.show()

# ---- Figure 5: Confusion Matrices ----
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
preds = [y_pred_dummy, y_pred_lr, y_pred_dt]
names = ['Dummy Baseline', 'Logistic Regression', 'Decision Tree']
for ax, pred, name in zip(axes, preds, names):
    cm = confusion_matrix(y_test, pred)
    disp = ConfusionMatrixDisplay(cm, display_labels=['Fail', 'Pass'])
    disp.plot(ax=ax, cmap='Blues', values_format='d')
    mf1 = f1_score(y_test, pred, average='macro')
    ax.set_title(f'{name}\nMacro F1 = {mf1:.3f}', fontsize=12, fontweight='bold')
plt.suptitle('Confusion Matrices for All Three Models', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig(f'{OUT}/fig5_confusion_matrices.png', dpi=150, bbox_inches='tight')
plt.show()

# ============================================================
# 7. FEATURE INTERPRETATION
# ============================================================
print("\n" + "=" * 70)
print("SECTION 7: FEATURE INTERPRETATION")
print("=" * 70)

print("\n--- 7a. Logistic Regression Coefficients ---")
lr_coefs = pd.DataFrame({
    'Feature': X_encoded.columns,
    'Coefficient': lr.coef_[0]
}).sort_values('Coefficient', key=abs, ascending=False)
print(lr_coefs.head(15).to_string(index=False))

# ---- Figure 6: LR Coefficients ----
fig, ax = plt.subplots(figsize=(10, 8))
top_lr = lr_coefs.head(15).sort_values('Coefficient')
colors_lr = ['#e74c3c' if v < 0 else '#2ecc71' for v in top_lr['Coefficient']]
ax.barh(top_lr['Feature'], top_lr['Coefficient'], color=colors_lr, edgecolor='black')
ax.set_xlabel('Coefficient Value', fontsize=12)
ax.set_title('Top 15 Logistic Regression Coefficients', fontsize=14, fontweight='bold')
ax.axvline(x=0, color='black', linewidth=0.8)
plt.tight_layout()
plt.savefig(f'{OUT}/fig6_lr_coefficients.png', dpi=150, bbox_inches='tight')
plt.show()

print("\n--- 7b. Decision Tree Feature Importance ---")
dt_importance = pd.DataFrame({
    'Feature': X_encoded.columns,
    'Importance': dt.feature_importances_
}).sort_values('Importance', ascending=False)
print(dt_importance[dt_importance['Importance'] > 0].to_string(index=False))

# ---- Figure 7: DT Feature Importance ----
fig, ax = plt.subplots(figsize=(10, 8))
top_dt = dt_importance[dt_importance['Importance'] > 0].sort_values('Importance')
ax.barh(top_dt['Feature'], top_dt['Importance'], color='#3498db', edgecolor='black')
ax.set_xlabel('Gini Importance', fontsize=12)
ax.set_title('Decision Tree Feature Importance', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig(f'{OUT}/fig7_dt_feature_importance.png', dpi=150, bbox_inches='tight')
plt.show()

# ---- Figure 8: Decision Tree Visualisation ----
fig, ax = plt.subplots(figsize=(24, 12))
plot_tree(dt, feature_names=X_encoded.columns.tolist(),
          class_names=['Fail', 'Pass'], filled=True, rounded=True,
          ax=ax, fontsize=8, max_depth=3)
ax.set_title('Decision Tree Structure (max depth 3 shown)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig(f'{OUT}/fig8_decision_tree.png', dpi=150, bbox_inches='tight')
plt.show()

# --- 7c. SHAP Analysis ---
print("\n--- 7c. SHAP Analysis ---")

explainer_dt = shap.TreeExplainer(dt)
shap_values_dt = explainer_dt.shap_values(X_test)

if isinstance(shap_values_dt, list):
    shap_vals = shap_values_dt[1]
else:
    shap_vals = shap_values_dt

# ---- Figure 9: SHAP Summary Plot (Decision Tree) ----
fig, ax = plt.subplots(figsize=(10, 8))
shap.summary_plot(shap_vals, X_test, feature_names=X_encoded.columns.tolist(),
                  show=False, max_display=15)
plt.title('SHAP Summary Plot (Decision Tree - Pass Class)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig(f'{OUT}/fig9_shap_summary_dt.png', dpi=150, bbox_inches='tight')
plt.show()

# ---- Figure 10: SHAP Bar Plot ----
fig, ax = plt.subplots(figsize=(10, 8))
shap.summary_plot(shap_vals, X_test, feature_names=X_encoded.columns.tolist(),
                  plot_type='bar', show=False, max_display=15)
plt.title('Mean |SHAP| Values (Decision Tree)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig(f'{OUT}/fig10_shap_bar_dt.png', dpi=150, bbox_inches='tight')
plt.show()

# SHAP for Logistic Regression
print("\nComputing SHAP for Logistic Regression (sample of 100)...")
X_test_scaled_df = pd.DataFrame(X_test_scaled, columns=X_encoded.columns)
background = shap.kmeans(X_test_scaled_df, 10)
explainer_lr = shap.KernelExplainer(lr.predict_proba, background)
shap_values_lr = explainer_lr.shap_values(X_test_scaled_df.iloc[:100])

if isinstance(shap_values_lr, list):
    shap_vals_lr = shap_values_lr[1]
else:
    shap_vals_lr = shap_values_lr

# ---- Figure 11: SHAP Summary (Logistic Regression) ----
fig, ax = plt.subplots(figsize=(10, 8))
shap.summary_plot(shap_vals_lr, X_test_scaled_df.iloc[:100],
                  feature_names=X_encoded.columns.tolist(),
                  show=False, max_display=15)
plt.title('SHAP Summary Plot (Logistic Regression - Pass Class)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig(f'{OUT}/fig11_shap_summary_lr.png', dpi=150, bbox_inches='tight')
plt.show()

# ============================================================
# 8. FAIRNESS AND PROXY RISK ANALYSIS
# ============================================================
print("\n" + "=" * 70)
print("SECTION 8: FAIRNESS AND PROXY RISK ANALYSIS")
print("=" * 70)

protected_attrs = ['sex', 'address']

for attr in protected_attrs:
    print(f"\n--- Fairness Analysis: {attr} ---")
    groups = X_test[attr].unique()

    for model_name, preds in [('Logistic Regression', y_pred_lr), ('Decision Tree', y_pred_dt)]:
        print(f"\n  {model_name}:")
        group_metrics = {}
        for g in sorted(groups):
            mask = X_test[attr] == g
            if mask.sum() == 0:
                continue
            acc = accuracy_score(y_test[mask], preds[mask])
            f1_g = f1_score(y_test[mask], preds[mask], average='macro')
            tpr = recall_score(y_test[mask], preds[mask], pos_label=1)
            tn = ((y_test[mask] == 0) & (preds[mask] == 0)).sum()
            fp = ((y_test[mask] == 0) & (preds[mask] == 1)).sum()
            fpr = fp / (fp + tn) if (fp + tn) > 0 else 0
            ppr = (preds[mask] == 1).mean()

            if attr == 'sex':
                label = 'Female' if g == 0 else 'Male'
            elif attr == 'address':
                label = 'Rural' if g == 0 else 'Urban'
            else:
                label = str(g)

            group_metrics[label] = {'Accuracy': acc, 'Macro F1': f1_g,
                                     'TPR': tpr, 'FPR': fpr, 'PPR': ppr}
            print(f"    {label}: Acc={acc:.3f}, F1={f1_g:.3f}, TPR={tpr:.3f}, FPR={fpr:.3f}, PPR={ppr:.3f}")

        if len(group_metrics) == 2:
            labels = list(group_metrics.keys())
            tpr_ratio = min(group_metrics[labels[0]]['TPR'], group_metrics[labels[1]]['TPR']) / \
                        max(group_metrics[labels[0]]['TPR'], group_metrics[labels[1]]['TPR']) \
                        if max(group_metrics[labels[0]]['TPR'], group_metrics[labels[1]]['TPR']) > 0 else 0
            ppr_ratio = min(group_metrics[labels[0]]['PPR'], group_metrics[labels[1]]['PPR']) / \
                        max(group_metrics[labels[0]]['PPR'], group_metrics[labels[1]]['PPR']) \
                        if max(group_metrics[labels[0]]['PPR'], group_metrics[labels[1]]['PPR']) > 0 else 0
            print(f"    Equal Opportunity Ratio (TPR): {tpr_ratio:.3f} (>=0.8 is fair)")
            print(f"    Statistical Parity Ratio (PPR): {ppr_ratio:.3f} (>=0.8 is fair)")

# ---- Figure 12: Fairness by Gender ----
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
for idx, (model_name, preds) in enumerate([('Logistic Regression', y_pred_lr), ('Decision Tree', y_pred_dt)]):
    metrics_by_group = {}
    for g in sorted(X_test['sex'].unique()):
        mask = X_test['sex'] == g
        label = 'Female' if g == 0 else 'Male'
        metrics_by_group[label] = {
            'Accuracy': accuracy_score(y_test[mask], preds[mask]),
            'Macro F1': f1_score(y_test[mask], preds[mask], average='macro'),
            'TPR': recall_score(y_test[mask], preds[mask], pos_label=1),
            'PPR': (preds[mask] == 1).mean()
        }
    df_fair = pd.DataFrame(metrics_by_group).T
    df_fair.plot(kind='bar', ax=axes[idx], rot=0, edgecolor='black')
    axes[idx].set_title(f'{model_name}: Fairness by Gender', fontsize=12, fontweight='bold')
    axes[idx].set_ylim(0, 1.1)
    axes[idx].axhline(y=0.8, color='red', linestyle='--', alpha=0.5, label='Fairness threshold')
    axes[idx].legend(fontsize=8)
plt.suptitle('Fairness Metrics Across Gender Groups', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig(f'{OUT}/fig12_fairness_gender.png', dpi=150, bbox_inches='tight')
plt.show()

# ---- Figure 13: Fairness by Address ----
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
for idx, (model_name, preds) in enumerate([('Logistic Regression', y_pred_lr), ('Decision Tree', y_pred_dt)]):
    metrics_by_group = {}
    for g in sorted(X_test['address'].unique()):
        mask = X_test['address'] == g
        label = 'Rural' if g == 0 else 'Urban'
        metrics_by_group[label] = {
            'Accuracy': accuracy_score(y_test[mask], preds[mask]),
            'Macro F1': f1_score(y_test[mask], preds[mask], average='macro'),
            'TPR': recall_score(y_test[mask], preds[mask], pos_label=1),
            'PPR': (preds[mask] == 1).mean()
        }
    df_fair = pd.DataFrame(metrics_by_group).T
    df_fair.plot(kind='bar', ax=axes[idx], rot=0, edgecolor='black')
    axes[idx].set_title(f'{model_name}: Fairness by Address', fontsize=12, fontweight='bold')
    axes[idx].set_ylim(0, 1.1)
    axes[idx].axhline(y=0.8, color='red', linestyle='--', alpha=0.5, label='Fairness threshold')
    axes[idx].legend(fontsize=8)
plt.suptitle('Fairness Metrics Across Urban/Rural Groups', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig(f'{OUT}/fig13_fairness_address.png', dpi=150, bbox_inches='tight')
plt.show()

# --- Proxy Risk Analysis ---
print("\n--- Proxy Risk Analysis ---")
non_protected = X_encoded.drop(columns=['sex', 'address'])
sex_corr = non_protected.corrwith(X_encoded['sex']).abs().sort_values(ascending=False)
addr_corr = non_protected.corrwith(X_encoded['address']).abs().sort_values(ascending=False)

print("\nCorrelation of features with 'sex':")
for feat, val in sex_corr.head(10).items():
    print(f"  {feat}: {val:.3f}")

print("\nCorrelation of features with 'address':")
for feat, val in addr_corr.head(10).items():
    print(f"  {feat}: {val:.3f}")

# ---- Figure 14: Proxy Correlation Heatmap ----
fig, ax = plt.subplots(figsize=(12, 6))
proxy_data = X_encoded.corr()[['sex', 'address']].drop(['sex', 'address'])
proxy_data = proxy_data.reindex(proxy_data.abs().max(axis=1).sort_values(ascending=False).head(15).index)
sns.heatmap(proxy_data, annot=True, fmt='.2f', cmap='RdBu_r', center=0, ax=ax,
            linewidths=0.5, vmin=-0.5, vmax=0.5)
ax.set_title('Top Features Correlated with Protected Attributes\n(Potential Proxy Variables)',
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig(f'{OUT}/fig14_proxy_correlation.png', dpi=150, bbox_inches='tight')
plt.show()

# ============================================================
# 9. SUMMARY TABLE
# ============================================================
print("\n" + "=" * 70)
print("SECTION 9: RESULTS SUMMARY")
print("=" * 70)

summary = pd.DataFrame({
    'Model': ['Dummy (Most Frequent)', 'Logistic Regression', 'Decision Tree'],
    'Test Accuracy': [
        accuracy_score(y_test, y_pred_dummy),
        accuracy_score(y_test, y_pred_lr),
        accuracy_score(y_test, y_pred_dt)
    ],
    'Test Macro F1': [macro_f1_dummy, macro_f1_lr, macro_f1_dt],
    'CV Mean Macro F1': [cv_results[m].mean() for m in models.keys()],
    'CV Std': [cv_results[m].std() for m in models.keys()],
    'Precision (Pass)': [
        precision_score(y_test, y_pred_dummy, pos_label=1),
        precision_score(y_test, y_pred_lr, pos_label=1),
        precision_score(y_test, y_pred_dt, pos_label=1)
    ],
    'Recall (Pass)': [
        recall_score(y_test, y_pred_dummy, pos_label=1),
        recall_score(y_test, y_pred_lr, pos_label=1),
        recall_score(y_test, y_pred_dt, pos_label=1)
    ]
})
print(summary.to_string(index=False))
summary.to_csv(f'{OUT}/results_summary.csv', index=False)
print("\nSaved: results_summary.csv")

# ---- Figure 15: Key Feature Distributions ----
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
key_features = ['failures', 'Medu', 'Fedu', 'studytime', 'absences', 'age']
for ax, feat in zip(axes.flatten(), key_features):
    for outcome_val, color, label in [(0, '#e74c3c', 'Fail'), (1, '#2ecc71', 'Pass')]:
        subset = df[df['outcome'] == outcome_val]
        ax.hist(subset[feat], bins=15, alpha=0.6, color=color, label=label, edgecolor='black')
    ax.set_title(f'{feat} Distribution by Outcome', fontsize=11, fontweight='bold')
    ax.set_xlabel(feat)
    ax.set_ylabel('Count')
    ax.legend()
plt.suptitle('Key Feature Distributions by Pass/Fail Outcome', fontsize=14, fontweight='bold', y=1.01)
plt.tight_layout()
plt.savefig(f'{OUT}/fig15_feature_distributions.png', dpi=150, bbox_inches='tight')
plt.show()

print("\n" + "=" * 70)
print("ALL OUTPUTS GENERATED SUCCESSFULLY")
print("=" * 70)
print(f"\nOutput files in: {OUT}/")
for f in sorted(os.listdir(OUT)):
    print(f"  {f}")
```
